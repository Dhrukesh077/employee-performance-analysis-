## 📚 Import Required Libraries

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['figure.figsize']=(8,5)

print("Libraries Imported Successfully.")
```

## 📂 Load Dataset

```
df = pd.read_csv("employee_performance.csv")

print("Shape:", df.shape)
df.head()
```

## 🔍 Dataset Information

```
df.info()
```

```
df.describe()
```

### Observation
- The dataset contains employee information.
- Numerical columns are used for statistical analysis.
- Categorical columns are used for probability analysis.


# ✅ Step 1 : Central Tendency & Dispersion

```
mean_salary = df['Salary'].mean()
median_salary = df['Salary'].median()
mode_salary = df['Salary'].mode()

variance_projects = df['Projects_Completed'].var()
std_projects = df['Projects_Completed'].std()

print("Mean Salary :", mean_salary)
print("Median Salary :", median_salary)
print("\nMode Salary")
print(mode_salary)

print("\nVariance of Projects Completed :", variance_projects)
print("Standard Deviation :", std_projects)
```

### Interpretation
- **Mean** represents the average salary.
- **Median** is the middle salary.
- **Mode** is the most frequent salary.
- **Variance** measures spread.
- **Standard Deviation** indicates average deviation from the mean.


# ✅ Step 2 : Probability

```
promotion_probability = (df['Promotion_Status']=="Yes").mean()

print("P(Promotion) =", promotion_probability)

contingency = pd.crosstab(df['Promotion_Status'],df['Department'])
print("\nContingency Table")
display(contingency)

high=df[df['Performance_Score']>80]
conditional=(high['Promotion_Status']=="Yes").mean()

print("\nP(Promotion | Performance Score > 80) =",conditional)
```

### Interpretation
Employees with higher performance generally have a greater probability of promotion.


# ✅ Step 3 : Visualization

```
# Histogram
plt.hist(df['Performance_Score'],bins=20,density=True)

mu=df['Performance_Score'].mean()
sigma=df['Performance_Score'].std()

x=np.linspace(df['Performance_Score'].min(),
              df['Performance_Score'].max(),200)

plt.plot(x,stats.norm.pdf(x,mu,sigma),linewidth=2)

plt.title("Histogram with Gaussian Curve")
plt.xlabel("Performance Score")
plt.ylabel("Density")
plt.show()
```

```
print("Skewness :",stats.skew(df['Salary']))
print("Kurtosis :",stats.kurtosis(df['Salary']))
```

```
stats.probplot(df['Projects_Completed'],dist="norm",plot=plt)
plt.title("Q-Q Plot")
plt.show()
```

### Interpretation
- Histogram helps visualize data distribution.
- Gaussian curve compares the data with a normal distribution.
- Skewness shows asymmetry.
- Kurtosis measures tail heaviness.
- Q-Q Plot checks normality.


# ✅ Step 4 : Linear Algebra

```
vectors=df[['Projects_Completed','Working_Hours']].head(5)
display(vectors)

v1=vectors.iloc[0].values
v2=vectors.iloc[1].values

dot=np.dot(v1,v2)
norm1=np.linalg.norm(v1,1)
norm2=np.linalg.norm(v1,2)

angle=np.degrees(
    np.arccos(dot/(np.linalg.norm(v1)*np.linalg.norm(v2)))
)

print("Dot Product :",dot)
print("L1 Norm :",norm1)
print("L2 Norm :",norm2)
print("Angle Between Vectors :",angle)
```
